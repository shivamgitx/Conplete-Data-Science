## Generators

Generators are a simple way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(3):
        return i**2
```

```python
square(3)
```

```python
##using yield
def square(n):
    for i in range(3):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
    
    ##this is a generator
```

```python
a=square(3)
a
```

```python
next(a)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
    yield 4
```

```python
gen=my_generator()
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
    
    ##this is a generator
```

Practical Example: Reading Large Files
Generators are particularly useful for reading large files because they allow you to process one line at a time without loading a entire file without loading the entire file in memory.

```python
### Practical:Reading Large file

def read_large_file(file_path):
    with open(file_path,'r')as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
    
    ##this is a generator
```
