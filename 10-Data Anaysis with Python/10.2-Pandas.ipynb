Pandas-DataFrame and Series

Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
#series
# pandas is one dimensional rray like object hat can holds any data type . it is similar to a column type.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("Series:\n",series)
print(type(series))
```

```python
# reate a series from dictionary

data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30,40,50]
index=['a','b','c','d','e']
pd.Series(data,index=index)
```

```python
## Dataframe
## create  dataframe form a dictuonary of list

data={
    'Name':['Shivam','John','Jack'],
    'Age':[23,30,45],
    'City':['Banglore','New York','Florida']
    }

df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a Dataframe frpma list of dictionary

data=[
    {'Name':'Shivam','Age':23,'City':'Banglore'},
     {'Name':'John','Age':20,'City':'New York'},
      {'Name':'Jack','Age':34,'City':'Florida'},
       {'Name':'Jesica','Age':36,'City':'Dubai'}, 
]

df=pd.DataFrame(data)
print(df)
```
