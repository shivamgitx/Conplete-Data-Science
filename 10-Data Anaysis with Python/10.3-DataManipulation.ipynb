## Data Manipulation and analysis with Pandas

Data manipulation and analysis are key tasks in any data science or data analysis project. Pandas provides a wide range of functions for data manipulation and analysis, making it easier to clean, transform, and extract insights from data. In this lesson, we will cover various data manipulation and analysis techniques using Pandas.

```python
import pandas as pd
```

```python
## Handling Missing Values

df.isnull()     #false means no missing values
                #Trues means missing values 
```

```python
df.isnull().sum()
```

```python
df.isnull().any()    #checks for any missing values
```

```python
df_filled=df.fillna(0)    #it fills missing values with 0
df
```

```python
### Filling missing value with the mean of the column
df['Sales_fillNA']=df['Sales'].fillna(df['Sales'].mean())
df
```

```python
df.dtypes
```

```python
## Renaming Columns
df=df.rename(columns={'Date':'Sales_Date'})
df.head()
```

```python
## change Datatypes
df['Value_new']=df['Value'].fillna(df['Value'].mean()).astype(int)         #you have to fill na values first
df.head()
```

```python
df['New_Value']=df['Value'].apply(lambda x: x*2)
df.head()
```

```python
## Data Aggregate and Grouping

grouped_mean=df.groupby('Category')['Value'].mean()
print(grouped_mean)
```

```python
df.groupby(['Product','Region'])['Value'].sum()      # total sale based on product and region
```

```python
df.groupby(['Product','Region'])['Value'].mean() 
```

```python
## multiple aggregate function

grouped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
## Merging and joining dataframe
#create sample dataframe
df1=pd.DataFrame({'Key': ['A','B','C'], 'Value': [1,2,3]})
df2=pd.DataFrame({'Key': ['A','B','D'], 'Value': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
# merge two dataframe on the 'key column'

pd.merge(df1,df2,on='Key',how='inner')
```

```python
pd.merge(df1,df2,on='Key',how='outer')
```

```python
pd.merge(df1,df2,on='Key',how='left')
```

```python
pd.merge(df1,df2,on='Key',how='right')
```
